## Part V – Analysis and Interpretation

**Q1. Vectorized operations, broadcasting, and element-wise computations in NumPy**

NumPy executes operations on entire arrays at once using optimized C-level routines, eliminating the overhead of Python `for` loops. For large datasets this can yield **10–100× speedups** because the computation stays within contiguous memory blocks (cache-friendly) and leverages BLAS/LAPACK routines. Broadcasting further avoids creating unnecessary copies of data—when a scalar or smaller array is combined with a larger one, NumPy *virtually* expands the smaller shape without allocating extra memory, keeping both time and space complexity low.

**Q2. Numerical issues in matrix operations**

Common issues include **floating-point round-off errors**, **catastrophic cancellation** (subtraction of nearly equal numbers), and **ill-conditioned matrices** (high condition number) where small input perturbations cause large output errors. For instance, a nearly singular matrix can make `np.linalg.det()` return a value close to zero that is numerically unreliable. Mitigation strategies include using `np.linalg.lstsq` for overdetermined systems, checking `np.linalg.cond()` before inversion, and using `np.float64` (or higher) precision.

**Q3. Advanced indexing – Boolean indexing and slicing**

Boolean indexing lets you select elements based on a condition (e.g., `arr[arr > 50]`), enabling expressive, readable filters without loops. Slicing provides O(1) *views* into arrays, so large arrays are not copied. Common **pitfalls**: (a) modifying a slice modifies the original—use `.copy()` to avoid unintended side effects; (b) chained indexing on Pandas DataFrames (e.g., `df[mask]['col']`) can produce a SettingWithCopyWarning and silently fail to update data—use `.loc[mask, 'col']` instead; (c) confusing `&` with `and` for Boolean arrays raises a ValueError.

**Q4. Reshape, flatten, transpose, and broadcasting – memory and performance**

Reshape and transpose return *views* (no data copy) whenever the memory layout allows, making them O(1) in both time and space. Flatten always returns a copy, and ravel returns a view when possible. Broadcasting avoids materializing expanded arrays, so `arr + scalar` does not allocate a full-size scalar array. The main implication is that **strides** (step sizes in bytes) change with transpose/reshape, and non-contiguous strides can slow certain operations. Using `np.ascontiguousarray()` can restore cache-friendly layout when performance degrades after repeated transposes.

**Q5. Statistical and aggregation operations in Pandas**

`groupby`, `corr`, `mean`, and `median` efficiently summarize large datasets and reveal trends such as salary growth by experience or education level. They scale well with chunked or categorically encoded data. **Limitations**: (a) `corr()` measures only *linear* associations—non-linear relationships (e.g., exponential salary curves) are missed unless the data is transformed first; (b) `groupby` with many small groups can be slow due to split-apply-combine overhead; (c) in high-dimensional data, pairwise correlation matrices suffer from the *curse of dimensionality*—most pairs appear uncorrelated even when complex interactions exist. Dimensionality reduction (PCA) or non-linear methods (mutual information, Spearman rank) should complement Pandas stats.

### Exercise 1 – Programming Language Growth Analysis (Python)

```python
import numpy as np
import pandas as pd

STUDENT_NAME  = 'TUPM-25-0586'
STUDENT_ID    = 250586
last3         = 586

# ── Load dataset ──────────────────────────────────────────────────────────
df_raw = pd.read_csv('programming_languages_popularity.csv')
df_raw['Month'] = pd.to_datetime(df_raw['Month'])
print('Columns:', df_raw.columns.tolist())
print(df_raw.head())

# ── Language assignment ────────────────────────────────────────────────────
lang_cols  = sorted([c for c in df_raw.columns if c != 'Month'])
lang_index = last3 % len(lang_cols)
assigned   = lang_cols[lang_index]
print(f'\nAssigned Language: {assigned}  (index {lang_index} of {lang_cols})')

# ── Extract time-series ────────────────────────────────────────────────────
df_lang = df_raw[['Month', assigned]].copy()
df_lang.columns = ['Month', 'Popularity']
df_lang.sort_values('Month', inplace=True)
df_lang.reset_index(drop=True, inplace=True)

# ── Growth metrics ─────────────────────────────────────────────────────────
df_lang['Growth_Rate'] = df_lang['Popularity'].pct_change() * 100
df_lang['Moving_Avg']  = df_lang['Popularity'].rolling(window=12).mean()
df_lang['Moving_STD']  = df_lang['Popularity'].rolling(window=12).std()

# ── Phase classification ───────────────────────────────────────────────────
growth_thresh  =  5.0
decline_thresh = -5.0
df_lang['Phase'] = np.select(
    [df_lang['Growth_Rate'] > growth_thresh,
     df_lang['Growth_Rate'] < decline_thresh],
    ['Growth', 'Decline'],
    default='Stable'
)

# ── Outputs ────────────────────────────────────────────────────────────────
print('\nDataFrame Output (first 10 rows):')
print(df_lang.head(10).to_string(index=False))

print('\nStatistical Summary:')
print(df_lang['Popularity'].describe())

print('\nGrowth Phase Counts:')
print(df_lang['Phase'].value_counts())

initial_pop = df_lang['Popularity'].iloc[0]
final_pop   = df_lang['Popularity'].iloc[-1]
overall_growth = ((final_pop - initial_pop) / initial_pop) * 100
print(f'\nKey Performance Metrics:')
print(f'  Initial Popularity : {initial_pop:.2f}%')
print(f'  Final Popularity   : {final_pop:.2f}%')
print(f'  Overall Growth     : {overall_growth:.2f}%')
print(f'  Growth periods     : {(df_lang["Phase"]=="Growth").sum()}')
print(f'  Decline periods    : {(df_lang["Phase"]=="Decline").sum()}')
print(f'  Stable periods     : {(df_lang["Phase"]=="Stable").sum()}')
```

### Exercise 2 – Lifecycle Classification of a Programming Language

```python
# ── Load & prepare ───────────────────────────────────────────────────────
df_raw2 = pd.read_csv('most-popular-programming-languages-2004-2024.csv')
df_raw2['Month'] = pd.to_datetime(df_raw2['Month'])

lang_cols2 = sorted([c for c in df_raw2.columns if c != 'Month'])
lang_index2 = last3 % len(lang_cols2)
assigned2   = lang_cols2[lang_index2]
print(f'Assigned Language: {assigned2}')

df_lc = df_raw2[['Month', assigned2]].copy()
df_lc.columns = ['Month', 'Popularity']
df_lc.sort_values('Month', inplace=True)
df_lc.reset_index(drop=True, inplace=True)

# ── Growth rate & smoothing ────────────────────────────────────────────────
df_lc['Growth_Rate'] = df_lc['Popularity'].pct_change() * 100
df_lc['Moving_Avg']  = df_lc['Popularity'].rolling(window=6).mean()
df_lc['Moving_STD']  = df_lc['Popularity'].rolling(window=6).std()

gr_mean = np.mean(df_lc['Growth_Rate'].dropna())
gr_std  = np.std(df_lc['Growth_Rate'].dropna())

# ── Lifecycle rules ────────────────────────────────────────────────────────
conditions = [
    df_lc['Growth_Rate'].abs() <= 1,
    df_lc['Growth_Rate'] > gr_mean,
    (df_lc['Growth_Rate'] > 0) & (df_lc['Growth_Rate'] <= gr_mean),
    df_lc['Growth_Rate'] < (-1 * gr_std)
]
choices = ['Maturity', 'Growth', 'Introduction', 'Decline']
df_lc['Lifecycle_Phase'] = np.select(conditions, choices, default='Stable')

print('\nLifecycle Classification Table (first 15 rows):')
print(df_lc.head(15).to_string(index=False))

counts = df_lc['Lifecycle_Phase'].value_counts()
pct    = (counts / len(df_lc) * 100).round(2)
print('\nPhase Counts and Percentages:')
print(pd.concat([counts, pct], axis=1, keys=['Count','%']))

dominant = counts.idxmax()
print(f'\nDominant Stage: {dominant}')
print(f'{assigned2} is currently in the {dominant.lower()} phase.')
```

```python
# ── Select two languages ─────────────────────────────────────────────────
lang_cols3 = sorted([c for c in df_raw2.columns if c != 'Month'])
idx_a = last3 % len(lang_cols3)
idx_b = (idx_a + 1) % len(lang_cols3)
lang_a, lang_b = lang_cols3[idx_a], lang_cols3[idx_b]
print(f'Language A: {lang_a}   Language B: {lang_b}')

df_cmp = df_raw2[['Month', lang_a, lang_b]].copy()
df_cmp.columns = ['Month', 'Lang_A', 'Lang_B']
df_cmp.sort_values('Month', inplace=True)
df_cmp.reset_index(drop=True, inplace=True)

# ── Central tendency ──────────────────────────────────────────────────────
for col, name in [('Lang_A', lang_a), ('Lang_B', lang_b)]:
    print(f'\n{name}  mean={df_cmp[col].mean():.3f}  '
          f'median={df_cmp[col].median():.3f}  '
          f'std={df_cmp[col].std():.3f}')

# ── Normalize ─────────────────────────────────────────────────────────────
a_vals = df_cmp['Lang_A'].values.astype(float)
b_vals = df_cmp['Lang_B'].values.astype(float)
a_norm = (a_vals - a_vals.mean()) / a_vals.std()
b_norm = (b_vals - b_vals.mean()) / b_vals.std()

# ── Correlation ────────────────────────────────────────────────────────────
corr_val = np.corrcoef(a_vals, b_vals)[0, 1]
print(f'\nCorrelation coefficient: {corr_val:.4f}')

# ── Dominance ratio ────────────────────────────────────────────────────────
dom_count = np.sum(a_vals > b_vals)
dom_ratio = dom_count / len(a_vals) * 100
print(f'Months {lang_a} > {lang_b}: {dom_count}  →  Dominance Ratio: {dom_ratio:.1f}%')

# ── Relative Performance Index ─────────────────────────────────────────────
rpi = (a_vals.mean() - b_vals.mean()) / b_vals.mean() * 100
print(f'Relative Performance Index (RPI): {rpi:.2f}%')

# ── Cross-over points ─────────────────────────────────────────────────────
cross = df_cmp['Month'][(a_vals > b_vals) & (np.roll(a_vals, 1) <= np.roll(b_vals, 1))]
print(f'\nCross-over points (months A surpasses B): {cross.values}')

# ── Summary DataFrame ─────────────────────────────────────────────────────
summary_cmp = pd.DataFrame([{
    'Mean_A'         : a_vals.mean(),
    'Mean_B'         : b_vals.mean(),
    'Std_A'          : a_vals.std(),
    'Std_B'          : b_vals.std(),
    'Correlation'    : corr_val,
    'Dominance_Ratio': dom_ratio,
    'RPI'            : rpi
}])
print('\nComparative Summary:')
print(summary_cmp.to_string(index=False))
```
